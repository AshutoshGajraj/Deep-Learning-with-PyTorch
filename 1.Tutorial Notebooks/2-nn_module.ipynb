```python
import torch
import torch.nn as nn
```

```python
class NNModel(nn.Module):
    def __init__(self, num_features): 
        super(NNModel,self).__init__()
        self.linear1 = nn.Linear(num_features, 3)
        self.relu = nn.ReLU()
        self.linear2 = nn.Linear(3, 1)
        self.sigmoid = nn.Sigmoid()

    def forward(self, features):
        x = self.linear1(features)
        x = self.relu(x)
        x = self.linear2(x)
        x = self.sigmoid(x)

        return x
```

```python
# creating features
X = torch.rand(5,5, dtype=torch.float32)
y = torch.tensor([1,0,0,1,1], dtype=torch.float32)
```

```python
# creating model object
model = NNModel(X.shape[1])
```

```python
# model summary
from torchinfo import summary

summary(model, X.shape)
```

```python
# cost function
criterion = nn.BCELoss()

# optimizer
optimizer = torch.optim.SGD(model.parameters(), lr=0.001)
```

```python
epochs = 100

for epoch in range(epochs):

    # forward pass
    y_pred = model(X)

    # loss calculation
    loss = criterion(y_pred, y.view(-1,1))

    # clear gradients to avoid accumulation
    optimizer.zero_grad()

    # backward pass
    loss.backward()

    # update parameters
    optimizer.step()

    if epoch % 10 == 0 or (epoch+1) % 10 == 0:
        print(loss)

```
