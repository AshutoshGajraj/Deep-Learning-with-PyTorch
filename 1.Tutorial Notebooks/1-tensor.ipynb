### Tensor Shapes

```python
x = torch.tensor([[3,2,4],[3,4,3],[5,5,3]])
print(x.shape)
```

```python
torch.empty_like(x) # for the same shape as x
```

```python
torch.zeros_like(x)
```

```python
torch.ones_like(x)
```

```python
torch.manual_seed(42)
torch.rand_like(x, dtype=torch.float64)
```

### Tensor Data Types

```python
# find data type
print(x.dtype)
```

```python
# assign dtype
torch.tensor([[1,2],[2,4]], dtype=torch.float32)
```

```python
torch.tensor([3.0,4.0], dtype=torch.int32)
```

```python
# using to()
print(x.dtype)

x = x.to(torch.float32)
print(x.dtype)
```

### Mathematical Operations

```python
# axis in numpy
import numpy as np

a = np.array([[1,2,3],[2,3,4]])
```

```python
print(a)
```

```python
print(np.sum(a, axis = 0))
```

```python
# dim in torch
a = torch.tensor([[1,2,3],[2,3,4]])
```

```python
print(torch.sum(a, dim=1))
```

```python
print(a)
```

```python
# argmax
torch.argmax(a)
```

```python
# transpose
print(a)

print()
print("Transposed:")
print(torch.transpose(a, 0, 1))

print()
print("Transposed:")
print(torch.transpose(a, 1, 0))
```

### Reshape Tensor


```python
a = torch.rand(2,3)
a
```

```python
# reshape
a.reshape(-1, 2)
```

```python
# flatten
a.flatten()
```

```python
a = torch.rand(2,3,4)
a
```

```python
# permute 
a = a.permute(2,0,1) # change dimensions 0,1,2 to 2,0,1
a.shape
```

```python
# unsqueeze
a = torch.rand(226,226,3)
a.unsqueeze(0).shape # add a dimension at 0th position of dimension
```

```python
# squeeze
a = torch.rand(2,1)
a.squeeze(1).shape
```

### Using GPU

```python
if torch.backends.mps.is_available():
  mps_device = torch.device('mps')
  print('MPS device found. Using Mac GPU.')
else:
  print('MPS device not found. Using CPU.')
```

```python
# creating tensor on GPU
torch.tensor([1,2,3], device=mps_device)
```

```python
# moving an existing tensor to GPU
b = a.to(device=mps_device)
b
```
